```python
import numpy as np
import matplotlib.pyplot as plt
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("warby_parker.db")  

df_funnel = pd.read_sql_query("""
WITH funnel AS (
  SELECT
    q.user_id,
    CASE WHEN h.user_id IS NOT NULL THEN 1 ELSE 0 END AS is_home_try_on,
    h.number_of_pairs,
    CASE WHEN p.user_id IS NOT NULL THEN 1 ELSE 0 END AS is_purchase
  FROM quiz q
  LEFT JOIN home_try_on h ON q.user_id = h.user_id
  LEFT JOIN purchase p ON q.user_id = p.user_id
)
SELECT * FROM funnel;
""", conn)
```

```python
n_quiz = len(df_funnel)
n_try = df_funnel["is_home_try_on"].sum()
n_buy = df_funnel["is_purchase"].sum()

quiz_to_try = n_try / n_quiz
try_to_buy = n_buy / max(n_try, 1)
quiz_to_buy = n_buy / n_quiz

pd.DataFrame({
    "metric": ["Quiz→Try-On", "Try-On→Purchase", "Quiz→Purchase"],
    "rate": [quiz_to_try, try_to_buy, quiz_to_buy]
})
```

```python
stages = ["Quiz", "Home Try-On", "Purchase"]
counts = [n_quiz, n_try, n_buy]

plt.figure()
plt.plot(stages, counts, marker="o")
plt.xlabel("Funnel stage")
plt.ylabel("Users")
plt.title("Warby Parker Funnel (counts)")
plt.show()
```

```python
ab = (df_funnel[df_funnel["is_home_try_on"] == 1]
      .groupby("number_of_pairs")
      .agg(users=("user_id","count"),
           purchases=("is_purchase","sum")))
ab["purchase_rate"] = ab["purchases"] / ab["users"]
ab
```

```python
plt.figure()
plt.bar(ab.index.astype(str), ab["purchase_rate"])
plt.xlabel("Number of pairs")
plt.ylabel("Purchase rate")
plt.title("Purchase rate by Home Try-On variant")
plt.show()
```

```python
# --- A/B analysis: 3 pairs vs 5 pairs (robust to "3 pairs"/"5 pairs" strings) ---

# 1) Keep only users who reached home try-on (since number_of_pairs is defined there)
df_try = df_funnel[df_funnel["is_home_try_on"] == 1].copy()

# 2) Clean number_of_pairs to an integer (handles "3 pairs", "5 pairs", or "3"/"5")
df_try["number_of_pairs"] = (
    df_try["number_of_pairs"]
      .astype(str)
      .str.extract(r"(\d+)", expand=False)   # pull the first number found
)

# Drop rows where we couldn't parse a number (just in case)
df_try = df_try[df_try["number_of_pairs"].notna()].copy()

df_try["number_of_pairs"] = df_try["number_of_pairs"].astype(int)

# 3) Aggregate A/B metrics
ab = (df_try
      .groupby("number_of_pairs", as_index=True)
      .agg(
          users=("user_id", "nunique"),
          purchases=("is_purchase", "sum")
      )
)

ab["purchase_rate"] = ab["purchases"] / ab["users"]
ab = ab.sort_index()

display(ab)

# 4) Pull rates safely (won't crash if one group is missing)
rate3 = ab["purchase_rate"].get(3, np.nan)
rate5 = ab["purchase_rate"].get(5, np.nan)

abs_lift = rate5 - rate3
rel_lift = abs_lift / rate3 if pd.notna(rate3) and rate3 != 0 else np.nan

print(f"3-pair purchase rate: {rate3:.3%}" if pd.notna(rate3) else "3-pair purchase rate: (missing)")
print(f"5-pair purchase rate: {rate5:.3%}" if pd.notna(rate5) else "5-pair purchase rate: (missing)")
print(f"Absolute lift (5 - 3): {abs_lift:.3%}" if pd.notna(abs_lift) else "Absolute lift: (cannot compute)")
print(f"Relative lift: {rel_lift:.2%}" if pd.notna(rel_lift) else "Relative lift: (cannot compute)")
```
